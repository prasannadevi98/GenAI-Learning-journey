## 1. Single-line comments

A single-line comment starts with `#`. Python ignores everything after `#` on that line.


```python
# Calculate the total price of two items
pen_price = 25
book_price = 80

# Add both prices
total_price = pen_price + book_price
print("Total price:", total_price)
```

## 2. Multi-line text using triple quotes

Triple quotes (`'''...'''` or `"""..."""`) create a multi-line string. They are also commonly used for documentation strings (docstrings).


```python
message = """Python is easy to read.
It is used in many fields.
Practice makes coding easier."""

print(message)
```

## 3. Python keywords

Keywords are reserved words with special meaning in Python. They cannot be used as normal variable names.


```python
import keyword

print("Number of Python keywords:", len(keyword.kwlist))
print("First 10 keywords:", keyword.kwlist[:10])

print("Is 'for' a keyword?", keyword.iskeyword("for"))
print("Is 'student' a keyword?", keyword.iskeyword("student"))
```

## 4. Python is case-sensitive

Uppercase and lowercase names are treated as different names. For example, `city` and `City` are two separate variables. Python constants such as `True`, `False`, and `None` also have exact capitalization.


```python
city = "Chennai"
City = "Madurai"

print("city =", city)
print("City =", City)
print("Are they the same value?", city == City)

print(True, False, None)
```

## 5. Variables and naming rules

A valid variable name:
- starts with a letter or underscore,
- can contain letters, numbers, and underscores after that,
- cannot contain spaces or symbols such as `%`,
- cannot be a Python keyword.

Examples of valid names: `student_name`, `_score`, `mark2`.

Examples of invalid names (shown only as text so the notebook remains runnable): `2mark`, `class`, `total%`, `student name`.


```python
student_name = "Arun"
_score = 92
mark2 = 88
course_name = "Python Basics"

print(student_name)
print(_score)
print(mark2)
print(course_name)
```

## 6. Multiple assignment

Python can assign multiple variables in one line. It can also assign the same value to several variables.


```python
# Different values to different variables
product, quantity, price = "Notebook", 3, 45
print(product, quantity, price)

# Same value to several variables
x = y = z = 10
print(x, y, z)
```

## 7. Integer data type (`int`)

Integers are whole numbers without a decimal point. They can be positive, negative, or zero.


```python
students = 32
temperature_change = -4
balance_items = 0

print(students, type(students))
print(temperature_change, type(temperature_change))
print(balance_items, type(balance_items))
```

## 8. Float data type (`float`)

Floats are numbers with decimal values. Scientific notation such as `3.2e4` is also stored as a float.


```python
height = 5.75
discount = -12.5
population_sample = 3.2e4

print(height, type(height))
print(discount, type(discount))
print(population_sample, type(population_sample))
```

## 9. Complex data type (`complex`)

A complex number contains a real part and an imaginary part. Python uses `j` for the imaginary component.


```python
signal = 4 + 7j

print("Complex value:", signal)
print("Real part:", signal.real)
print("Imaginary part:", signal.imag)
print("Type:", type(signal))
```

## 10. String data type (`str`)

Strings store text. They can use single quotes or double quotes. Using `+` with strings joins them together.


```python
first_name = "Meena"
last_name = 'Kumar'
full_name = first_name + " " + last_name

print(full_name)
print("Type:", type(full_name))

# String concatenation does not perform arithmetic
number_text = "20" + "30"
print("String result:", number_text)
```

## 11. Boolean data type (`bool`)

Booleans have only two values: `True` and `False`. They are commonly produced by comparisons.


```python
is_logged_in = True
is_raining = False
is_adult = 21 >= 18

print(is_logged_in, type(is_logged_in))
print(is_raining, type(is_raining))
print(is_adult, type(is_adult))
```

## 12. Checking data types with `type()`

The built-in `type()` function tells us the data type of a value or variable.


```python
values = [250, 18.75, 2-3j, "Python", False]

for value in values:
    print(value, "->", type(value))
```

## 13. Using `print()`

`print()` displays values on the screen. It can print text, calculations, and multiple items together.


```python
item_price = 120
quantity = 4

print("Item price:", item_price)
print("Quantity:", quantity)
print("Total:", item_price * quantity)
print("Python", "Day", 1, sep=" - ")
```

## 14. Escape sequences

Escape sequences begin with a backslash (`\`) and help format text.

- `\\` prints a backslash
- `\'` prints a single quote inside a single-quoted string
- `\"` prints a double quote inside a double-quoted string
- `\n` moves to a new line
- `\t` inserts a tab


```python
print("Folder path: C:\\Users\\Student")
print('She said, \'Python is useful\'')
print("He said, \"Keep practicing\"")
print("Line 1\nLine 2\nLine 3")
print("Name\tScore")
print("Ravi\t95")
```

## 15. Multi-line string and formatted output

A triple-quoted string can hold several lines exactly as written. This is useful for paragraphs, templates, and documentation.


```python
profile = """Student Profile
---------------
Name: Anitha
Course: Python
Level: Beginner"""

print(profile)
```

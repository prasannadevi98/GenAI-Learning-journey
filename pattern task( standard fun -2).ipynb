## ** print the blow pattern*

-----
 ----
  ---
   --
    -

```python
x = 5
for i in range(5):
    print(" " * i + "-" * (x - i))
print()
    
```

## ** print the blow pattern*

 *********
  *******
   *****
    ***
     *

```python
x = 9
for i in range((x + 1) // 2):
    print(" " * i + "*" * (x - 2 * i))
    
```
